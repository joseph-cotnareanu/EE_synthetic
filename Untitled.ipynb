```python
from datasets import load_dataset


alphabet = {'A':[1,0,0,0,0], 'B':[0,1,0,0,0], 'C':[0,0,1,0,0], 'D':[0,0,0,1,0], 'E':[0,0,0,0,1]}
alpha_num = {0: 'A', 1: 'B', 2:'C', 3:'D', 4: 'E'}
ds = load_dataset("hails/agieval-aqua-rat")
ds = ds['test']
```

```python

import json


aqua_training_path = '/home/joseph/EE_llm/aqua_train.json'
a_t = open(aqua_training_path, 'r',  encoding='utf-8-sig')
aqua_training = []
for line in a_t.readlines():
    datum=json.loads(line)
    tmp = {}
    tmp['query'] = datum['question']
    option_str = 'Answer Choices: '
    for option in datum['options']:
        option_str += '(' + option + ' '
    option_str = option_str[:-1] +'\n Among A through E, the answer is,'
    tmp['query'] += option_str
    tmp['gold'] = [datum['correct']]
    aqua_training.append(tmp)
```

```python
paths = ['/home/joseph/flo/EE_synthetic/s_cost0.0001.csv', '/home/joseph/flo/EE_synthetic/s_cost0.01.csv', '/home/joseph/flo/EE_synthetic/s_cost0.1.csv']
s = []
import csv
import numpy as np


for path in paths:
    c = csv.reader(open(path, 'r'))
    s.append([])
    for row in c:
        s[-1].append(int(row[0][0]))

ad = []
nd = []

for i in range(100):
    if s[0][i] == s[1][i] and s[1][i] == s[2][i] and s[0][i] == 1:
        ad.append(i)
    elif s[0][i] == s[1][i] and s[1][i] == s[2][i] and s[0][i] == 0:
        nd.append(i)   
        

```

```python
ap = []
np = []
for i in ad:
    ap.append(aqua_training[900+i])
for i in nd:
    np.append(aqua_training[900+i])
```

```python
import pickle as pkl
pkl.dump(ap,open('./always_deferred.pkl', 'wb'))
pkl.dump(np,open('./never_deferred.pkl', 'wb'))
```
